```python
import os
import glob
import zipfile

for filename in uploaded.keys():
    if filename.endswith('.zip'):
        with zipfile.ZipFile(filename, 'r') as z:
            z.extractall('/content/superstore')

csv_files = glob.glob('/content/**/*.csv', recursive=True)

print("Available CSV files:")
print(csv_files)
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'], errors='coerce'
)
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
```

```python
print("Total Sales:", round(df['Sales'].sum(), 2))
```

```python
state_sales = df.groupby('State')['Sales'].sum()
print(state_sales)
```

```python
top_states = state_sales.sort_values(
    ascending=False
).head(10)

print(top_states)
```

```python
fig = px.choropleth(
    state_map_data,
    locations='State_Code',
    locationmode='USA-states',
    color='Sales',
    scope='usa',
    color_continuous_scale='Blues',
    title='State-wise Sales Distribution'
)

fig.show()
```

```python
city_sales = df.groupby('City')['Sales'].sum().reset_index()
print(city_sales.head())
```

```python
city_sales = city_sales.sort_values(
    by='Sales', ascending=False
)
```

```python
top_cities = city_sales.head(10)
display(top_cities)
```

```python
plt.figure(figsize=(12,6))
plt.bar(top_cities['City'], top_cities['Sales'])
plt.title('Top 10 Cities by Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
city_orders = df.groupby('City')['Order ID'].nunique().reset_index()
city_orders.columns = ['City', 'Total Orders']
display(city_orders.head(10))
```

```python
top_demand_cities = city_orders.sort_values(
    by='Total Orders', ascending=False
).head(10)

display(top_demand_cities)
```

```python
plt.figure(figsize=(12,6))
plt.bar(top_demand_cities['City'],
        top_demand_cities['Total Orders'])
plt.title('Top 10 Cities by Order Demand')
plt.xlabel('City')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
city_analysis = pd.merge(
    city_sales,
    city_orders,
    on='City'
)

display(city_analysis.head())
```

```python
city_analysis['Average Sales Per Order'] = (
    city_analysis['Sales'] / city_analysis['Total Orders']
)

display(city_analysis.head(10))
```

```python
print("Highest Demand City:")
print(city_analysis.loc[
    city_analysis['Total Orders'].idxmax()
])
```

```python
total_orders = city_analysis['Total Orders'].sum()
print("Total Orders:", total_orders)
```

```python
total_city_sales = city_analysis['Sales'].sum()
print("Total City Sales:", round(total_city_sales, 2))
```

```python
city_analysis['Sales Contribution (%)'] = (
    city_analysis['Sales'] / total_city_sales * 100
)

display(city_analysis.head(10))
```

```python
high_demand = city_analysis[
    city_analysis['Total Orders'] >=
    city_analysis['Total Orders'].quantile(0.75)
]

display(high_demand)
```

```python
low_demand = city_analysis[
    city_analysis['Total Orders'] <=
    city_analysis['Total Orders'].quantile(0.25)
]

display(low_demand)
```

```python
expansion_cities = high_demand.sort_values(
    by='Sales', ascending=False
)

display(expansion_cities.head(10))
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    expansion_cities['City'].head(10),
    expansion_cities['Total Orders'].head(10)
)

plt.title('Potential Expansion Cities')
plt.xlabel('City')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
fig = px.scatter(
    city_analysis,
    x='Total Orders',
    y='Sales',
    hover_name='City',
    title='City-wise Sales vs Customer Demand',
    size='Sales'
)

fig.show()
```

```python
top_expansion = expansion_cities.head(5)

display(top_expansion)
```

```python
print("BUSINESS EXPANSION ANALYSIS")
print("---------------------------")

for city in top_expansion['City']:
    print("Potential City:", city)

print("High order volume indicates stronger customer demand.")
```
